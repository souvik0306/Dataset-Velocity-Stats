# ROS Bag Motion Metrics

This notebook reads ROS1 bag files and computes the requested velocity, acceleration, angular-rate, roll, and pitch metrics only inside the configured evaluation windows. It uses `rosbag_motion_metrics.py` for the actual calculations so the command-line script and notebook stay consistent.

```python
from pathlib import Path
import pandas as pd

from rosbag_motion_metrics import compute_metrics_for_bag, find_bags, flatten_metric_rows

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)
```

## Configure Input and Output

```python
# Point this at a single .bag file or a directory containing .bag files.
ROOT = Path("2026-07-31/RAW/flight_1.bag")
OUTPUT_CSV = Path("rosbag_motion_metrics_first_bag.csv")

bags = find_bags(ROOT)
print(f"Found {len(bags)} bag file(s)")
bags[:5]
```

## Compute Metrics

```python
wide_rows = []
for index, bag_path in enumerate(bags, start=1):
    print(f"[{index}/{len(bags)}] {bag_path}")
    wide_rows.append(compute_metrics_for_bag(bag_path))

metrics = pd.DataFrame(flatten_metric_rows(wide_rows))
metrics
```

## Save Results

```python
metrics.to_csv(OUTPUT_CSV, index=False)
OUTPUT_CSV
```
